## .Customer Segmentation: 
### Utilize clustering algorithms (e.g., K-means) to segment customers based on behavior and purchase patterns.

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
```

```python
csa = pd.read_csv('ifood_df.csv')
```

```python
csa.head()
```

```python
features = ['Income','MntTotal']
customer_data = csa[features].dropna()
```

```python
# SCALE THE DATA
# Income is in the tens-of-thousands, but MntTotal is in the hundreds.
# We have to "scale" them so the algorithm treats them equally.

scaler = StandardScaler()
scaled_data = scaler.fit_transform(customer_data)
```

```python
# Running the K-means algorithm
# we are telling it to make 3 Distinct customer group(clusters)

kmeans= KMeans(n_clusters= 3,random_state = 42)
customer_data['Customer_Segment'] = kmeans.fit_predict(scaled_data)

segment_summary = customer_data.groupby('Customer_Segment').mean().round(2)
segment_summary['Number_of_customers'] = customer_data['Customer_Segment'].value_counts()
print(segment_summary)
```

## .Visualization:
### Creating visualizations (e.g., scatter plots, bar charts) to illustrate customer segments.

```python
#  --------------- Scatter Plot ------------------------

import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10,6))
sns.scatterplot(data=customer_data, x='Income', y='MntTotal', hue='Customer_Segment', palette='Set1')
plt.title('Customer Profiles: Spending Behavior vs. Annual Income')
plt.show()
```

```python
# ----------------------Countplot ----------------------

plt.figure(figsize=(10,6))
sns.countplot(data=customer_data,x='Customer_Segment',palette= 'Set1')
plt.title('Customer_Segment Distribution')
plt.ylabel('Number of Customers')
plt.show()
```

## Customer Segmentation Analysis: Strategic Insights and Recommendations
### Overview
##### Utilizing K-Means clustering algorithms, the customer base was segmented into three distinct profiles based on annual income and total purchase value. This segmentation allows for highly targeted, data-driven marketing strategies designed to maximize revenue, improve retention, and optimize the customer experience for each unique demographic.

#### 1. Segment 1: The "Value-Conscious" Base (Blue Group)
#### Profile: Low Income (~$33,300) | Low Spend (~$100)
#### Segment Size: Largest market share

#### Insight: This segment represents the largest portion of the customer base but yields the lowest average order value. These consumers
####           are highly price-sensitive and strictly bargain-driven. They are unlikely to respond to standard marketing but are   highly motivated by significant, tangible deals that allow them to stretch their purchasing power.

#### Strategic Recommendation: Implement Volume-Based Discounting.

#### Introduce a tiered voucher system that rewards volume rather than single-item purchases (e.g., "Buy 4 items, receive 50% off").

#### This strategy effectively neutralizes bargain-hunting behaviors by gamifying the discount, thereby forcing an increase in cart size and overall transaction value without sacrificing baseline profit margins.

#### 2. Segment 2: The "Core Middle" Base (Green Group)
#### Profile: Moderate Income (~$61,300) | Moderate Spend (~$635)
#### Segment Size: Strong secondary market share

#### Insight: This demographic forms the reliable financial backbone of the business. They prioritize quality, longevity, and practical utility over luxury status or steep discounts. They are highly risk-averse regarding unexpected future costs (such as maintenance or replacement) and value long-term peace of mind.

#### Strategic Recommendation: Focus on Value-Add Promotions and Retention.

#### Shift promotional focus away from direct price cuts and toward risk-reduction offers, such as extended 5-year warranties, guaranteed return policies, or free maintenance periods.

#### Implement a robust points-based loyalty program to incentivize repeat purchases, maximizing Customer Lifetime Value (CLV) and preventing churn to market competitors.

#### 3. Segment 0: The "Premium" VIP Base (Red Group)
#### Profile: High Income (~$77,500) | High Spend (~$1,455)
#### Segment Size: Smallest, but most lucrative per capita

#### Insight: Despite being the smallest segment, these customers drive disproportionately high revenue. They possess high disposable income and do not factor standard discounts into their purchasing decisions. For this group, the perceived value lies in exclusivity, rarity, and the prestige of the shopping experience. Standard discount coupons actively risk devaluing the brand in their eyes.

#### Strategic Recommendation: Develop an Exclusive "Velvet Rope" Experience.

#### Establish an invite-only VIP tier that offers early access to new product drops or limited-edition items.

#### Elevate the physical presentation of products through luxury packaging, customized unboxing experiences, and priority concierge customer service. Market these items as curated, high-status experiences rather than retail transactions.
